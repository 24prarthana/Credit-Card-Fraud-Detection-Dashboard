Data preprocessing

```python
x=df.drop("Class",axis=1)
y=df['Class']
```

```python
x.head()
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,random_state=42,stratify=y)
```

```python
df=df.dropna(subset=["Class"])
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
scaler = StandardScaler()
```

```python
x_train_scaled = scaler.fit_transform(x_train)
x_test_scaled = scaler.transform(x_test)
```

```python
x_train_scaled[:5]
```

"""#Imbalance"""

```python
from imblearn.under_sampling import RandomUnderSampler
```

```python
rus = RandomUnderSampler(random_state=42)
```

```python
x_train_balanced, y_train_balanced = rus.fit_resample(
    x_train_scaled,
    y_train
)
```

```python
y_train_balanced.value_counts()
```

Model train

```python
from sklearn.linear_model import LogisticRegression
```

```python
model = LogisticRegression(max_iter=1000)
model.fit(x_train_balanced, y_train_balanced)
```

```python
y_pred = model.predict(x_test_scaled)
```

```python
from sklearn.metrics import confusion_matrix
cm = confusion_matrix(y_test,y_pred)
print(cm)
```

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score
)
```

```python
print("Accuracy :", accuracy_score(y_test, y_pred))
print("Precision:", precision_score(y_test, y_pred))
print("Recall   :", recall_score(y_test, y_pred))
print("F1 Score :", f1_score(y_test, y_pred))
```

```python
from sklearn.metrics import classification_report
```

```python
print(classification_report(y_test, y_pred))
```

```python
from sklearn.metrics import roc_auc_score
```

```python
y_prob = model.predict_proba(x_test_scaled)[:, 1]
```

```python
print("ROC-AUC:", roc_auc_score(y_test, y_prob))
```

```python
from imblearn.under_sampling import RandomUnderSampler
rus = RandomUnderSampler(random_state = 42, sampling_strategy = 0.1)
```

```python
x_train_v2, y_train_v2 = rus.fit_resample(x_train_scaled,y_train)
```

```python
y_train_v2.value_counts()
```

```python
from sklearn.ensemble import RandomForestClassifier
```

```python
model_v2 = RandomForestClassifier(random_state=42)
```

```python
model_v2.fit(x_train_v2,y_train_v2)
```

```python
y_pred_v2 = model_v2.predict(x_test_scaled)
```

```python
from sklearn.metrics import confusion_matrix
cm = confusion_matrix(y_test,y_pred_v2)
print(cm)
```

```python
print("\nClassification Report:")
print(classification_report(y_test, y_pred_v2))
```

```python
print("Accuracy :", accuracy_score(y_test, y_pred_v2))
print("Precision:", precision_score(y_test, y_pred_v2))
print("Recall   :", recall_score(y_test, y_pred_v2))
print("F1 Score :", f1_score(y_test, y_pred_v2))
```
